```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = np.random.normal(size=1000)
```

```python
EMILY_PAL = sns.color_palette("rocket", 10).as_hex()

sns.set_theme(style="whitegrid", context="talk")

mpl.rcParams.update({
    "figure.figsize": (10, 5),
    "figure.dpi": 150,
    "savefig.dpi": 300,
    "savefig.bbox": "tight",
    "savefig.pad_inches": 0.05,
    "figure.constrained_layout.use": True,

    "figure.facecolor": "#F7F7F7",
    "axes.facecolor":   "#F7F7F7",

    "font.family": "DejaVu Sans",
    "axes.titlesize": 18,
    "axes.labelsize": 14,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
    "legend.fontsize": 12,
    "axes.titlepad": 10,

    "lines.linewidth": 2.5,
    "lines.markersize": 6,

    "axes.grid": True,
    "grid.color": "#C6CED5",
    "grid.alpha": 0.35,
    "grid.linewidth": 0.8,
    "axes.axisbelow": True, # grid behind data

    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.spines.left": False,
    "axes.spines.bottom": True,
    "axes.edgecolor": "#AAB4BE",
    "axes.linewidth": 0.8,

    "xtick.direction": "out",
    "ytick.direction": "out",
    "xtick.major.size": 0,
    "ytick.major.size": 0,

    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "svg.fonttype": "none",
})
def prettify_axes(ax, title=None, xlabel=None, ylabel=None):
    """Default helper for consistent titles/labels"""
    if title: ax.set_title(title, loc="left", fontweight="normal")
    if xlabel: ax.set_xlabel(xlabel)
    if ylabel: ax.set_ylabel(ylabel)
    ax.margins(x=0.02)
    return ax

def save_for_slides(fig, filename_base):
    """Export helper: presentation-safe outputs"""
    fig.savefig(f"{filename_base}.png", transparent=False)
```

```python
x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)
fig, ax = plt.subplots()
ax.plot(x, y, label="sin(x)")
ax.grid()
ax.set_xlim(0.5, 5)
ax.set_ylim(-2, 2)
ax.set_xlabel("This is the x-axis")
ax.set_ylabel("This is the y-axis")
ax.set_title("My hundredth plot")
z = 2*np.cos(x) # assign to a new variable, z
ax.plot(x, z, linewidth=3.0, linestyle="--", label="2cos(x)")
ax.legend()
save_for_slides(fig, "my_first_plot")
```

```python
fig2, ax2 = plt.subplots()
ax2.plot(data, color=EMILY_PAL[0], linewidth=1.0)

fig3, ax3 = plt.subplots()
ax3.hist(data, color=EMILY_PAL[0], alpha=0.5)
ax3.hist(data, bins=np.arange(-3, 3.1, 0.1), color=EMILY_PAL[0], alpha=0.5)
save_for_slides(fig2, "my_second_plot")
```

```python
fig2, ax2 = plt.subplots()
ax2.plot(data, color=EMILY_PAL[0], linewidth=1.0)
prettify_axes(ax2, title="Plot 1 Title", xlabel="Plot 1 X axis", ylabel="Plot 1 Y axis")


fig3, ax3 = plt.subplots()
ax3.hist(data, density=True, alpha=0.5)
ax3.hist(data, bins=np.arange(-3, 3.1, 0.1), density=True, alpha=0.5, color=EMILY_PAL[0])
prettify_axes(ax3, title="Plot 2 Title", xlabel="Plot 2 X axis", ylabel="Plot 2 Y axis")
save_for_slides(fig3, "my_third_plot")
```
